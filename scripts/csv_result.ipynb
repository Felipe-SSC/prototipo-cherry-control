```python
# open csv
import os

import pandas as pd

# La deteccion vive en detecciones/, el notebook en scripts/.
CARPETA_DETECCIONES = os.path.join("..", "detecciones")

df = pd.read_csv(os.path.join(CARPETA_DETECCIONES, "clasificacion_20260916_164651.csv"))

df
```

```python
# métricas de calibre estimado:

calibre = df['calibre_mm']

calibre.describe()
```

```python
## distribución de clases

import matplotlib.pyplot as plt

# Contar la cantidad de cada clase
class_counts = df['tipo'].value_counts()

# Crear un gráfico de barras
#size adjustment
plt.figure(figsize=(5, 3))
plt.bar(class_counts.index, class_counts.values, color=['black', 'green', 'blue'])
plt.xlabel('Clase')
plt.ylabel('Cantidad')
plt.title('Distribución de Clases')
plt.show()
```

```python
# distribucion de calibre mm
import numpy as np
import scipy.stats as stats

calibres = df['calibre_mm']

# grafico de histograma
plt.figure(figsize=(5, 3))
plt.hist(calibres, bins=20, color='pink', edgecolor='red')
#add distribution curve with scipy
x = np.linspace(calibres.min(), calibres.max(), 100)
plt.plot(x, stats.norm.pdf(x, calibres.mean(), calibres.std())*len(calibres)*20/100, 'k--', linewidth=2)

plt.xlabel('Calibre (mm)')
plt.ylabel('Frecuencia')
plt.title('Distribución de Calibre')
plt.show()
```
